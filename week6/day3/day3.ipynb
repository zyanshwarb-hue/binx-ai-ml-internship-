## 3️⃣ Step 2 — Smoke-Test the App Locally 🧪

Before deploying anywhere, the app must run cleanly in an isolated way — this catches
missing-import or path errors early, when they're cheap to fix, rather than discovering
them after deployment fails on the cloud.

**Command used** (run this from a terminal, not this notebook cell — Streamlit apps run
as their own server process, not as notebook code):
```bash
streamlit run app.py
```

````python
import subprocess, time, requests

# Launch the app as a background process on a test port
proc = subprocess.Popen(
    ["streamlit", "run", "app.py", "--server.headless", "true", "--server.port", "8501"],
    stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL
)
time.sleep(6)  # give the server a moment to boot

try:
    response = requests.get("http://localhost:8501", timeout=5)
    print(f" App booted successfully — HTTP status: {response.status_code}")
except Exception as e:
    print(f" App failed to boot: {e}")
finally:
    proc.terminate()
````

**Result:** an HTTP 200 response confirms the app served its front page correctly with
no startup errors — the model loaded, `st.set_page_config` ran, and all the form widgets
rendered without exceptions.

## 4️⃣ Step 3 — Deploy to Streamlit Community Cloud 🚀

This part happens outside this notebook, directly on GitHub and streamlit.io, since it
requires the intern's own accounts and credentials. The steps:

1. **Push this folder to GitHub** — `app.py`, `requirements.txt`, and
   `tuned_cardiac_pipeline.joblib` must all be committed to the repo (inside
   `project/` or a dedicated `deployment/` folder), since Streamlit Cloud builds
   directly from the GitHub repo, not from a local machine.
2. Go to **[share.streamlit.io](https://share.streamlit.io)** and sign in with GitHub.
3. Click **"New app"**, select the repo (`binx-ai-ml-internship-`), the branch
   (`main` or a feature branch), and the exact path to `app.py`
   (e.g. `project/deployment/app.py`).
4. Click **Deploy**. Streamlit Cloud will install everything from `requirements.txt`
   and boot the app automatically — this takes a couple of minutes the first time.
5. Once live, copy the public URL (format: `https://<app-name>.streamlit.app`).

## 6️⃣ Reflection ✍️

requirements.txt needs exact package names because Streamlit Community Cloud builds the app
from a completely clean environment - it does not reuse anything installed on my local
machine. I actually ran into a direct example of this locally: when I first tried the smoke
test, I got a FileNotFoundError because the streamlit command wasn't recognized - it turned
out Streamlit wasn't installed in my local Python environment yet. That's exactly the failure
mode requirements.txt exists to prevent on the cloud: if a package the app imports isn't
listed there, the deployed build will fail the same way my local test did, except I won't
have a terminal in front of me to fix it interactively.

The local smoke test (HTTP 200) verified that the app's Python code has no import errors,
that the trained pipeline loads correctly, and that Streamlit can render the page and all
the form widgets without crashing on startup. What it does NOT verify is anything about the
cloud environment itself - whether the exact package versions in requirements.txt resolve
correctly on Streamlit's servers, whether the joblib file is actually committed to GitHub
(and not accidentally excluded by .gitignore), or whether the app behaves the same way when
multiple people access it at once. That's exactly why the post-deployment checklist re-tests
everything directly on the live URL instead of trusting the local result alone.
